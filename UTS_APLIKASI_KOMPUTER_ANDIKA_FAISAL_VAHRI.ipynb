```python
volume = [120, 185, 150, 110, 135, 160, 145, 125, 175, 190]
total = sum(volume)
rata_rata = sum(volume) / len(volume)
tertinggi = max(volume)
terendah = min(volume)

print("total:", total)
print("Rata-rata:", rata_rata)
print("Tertinggi:", tertinggi)
print("Terendah:", terendah)
```

```python
import numpy as np

volume =np.array([120, 185, 150, 110, 135, 160, 145, 125, 175, 190])
print("Total Volume Kendaraan:", np.sum(volume))
print("Rata-rata Volume Kendaraan:", np.mean(volume))
print("Volume Kendaraan Tertinggi:", np.max(volume))
print("Volume Kendaraan Terendah:", np.min(volume))
```
